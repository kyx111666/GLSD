# Official-response locked-result preflight

This notebook runs the preflight gate only after Drive mount. It replays sealed Native and locked GLSD-90 selections, but does not run H/G/L/GL selection or bootstrap. It creates a new timestamped `official_response_ablation/preflight_*` directory and never writes into sealed evidence. Upload the companion `official_response_component_ablation.py` before continuing.

```python
from google.colab import drive, files
drive.mount('/content/drive')
uploaded = files.upload()
name = 'official_response_component_ablation.py'
assert name in uploaded, f'Upload {name} from the patch directory.'
open('/content/' + name, 'wb').write(uploaded[name])
print('PATCH_UPLOAD = PASS')
```

```python
import subprocess, sys
from datetime import datetime, timezone
from pathlib import Path

base = Path('/content/drive/MyDrive/official_response_ablation')
base.mkdir(parents=True, exist_ok=True)
log = base / ('preflight_launcher_' + datetime.now(timezone.utc).strftime('%Y%m%dT%H%M%SZ') + '.log')
command = [sys.executable, '/content/official_response_component_ablation.py', '--setting', 'all', '--preflight-only', '--output-root', str(base)]
with log.open('w', encoding='utf-8') as handle:
    result = subprocess.run(command, stdout=handle, stderr=subprocess.STDOUT, text=True)
print('LOG =', log)
print(log.read_text(encoding='utf-8')[-4000:])
assert result.returncode == 0, 'A preflight or replay gate failed; inspect the new log. Locked evidence was not modified.'
print('OFFICIAL_RESPONSE_PREFLIGHT = PASS')
```
